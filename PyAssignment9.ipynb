```python
#Problem 1: Zipper_file()
def zipper_file(filename, strings):
    with open(filename, 'w') as file:
        for i in range(0, len(strings), 2):
            result = strings[i] + strings[i + 1]
            file.write(result + '\n')
    print("File created successfully")
    with open(filename, 'r') as file:
        print(file.read())
strings = ["PyAssignment9", "Completed", "Assignment", "Done"]
zipper_file("output.txt", strings)
```

```python
#Problem 2: Reformat_date()
def reformat_date(dates):
    months = {
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }
    for date in dates:
        month, day, year = date.split("/")
        print("Office Hours on", months[month], day + ",", year)
dates = ["05/29/2025", "03/12/2025", "11/09/2025"]
reformat_date(dates)
```
