```python
# Execise 1: Handling ZeroDivisionError
def safe_divide(numerator, denominator):
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by Zero.")
        return None


numerator = float(input("Enter the numerator: "))
denominator = float(input("Enter the denominator: "))

result = safe_divide(numerator, denominator)
print("Result:", result)
```

```python
# Exercise 2 Handling ValueError
import math

def square_root(number1):
    try:
        if number1 < 0:
            raise ValueError
        return math.sqrt(number1)
    except ValueError:
        print("Invalid input! Please enter a positive integer or a float value.")
        return None


number1 = float(input("Enter a number: "))

result = square_root(number1)
print("Square Root:", result)
```

```python
# Exercise 3 Handling Generic Exceptions
def calculate(num):
    try:
        result = num / (num - 5)
        return result
    except Exception as e:
        print("An error occurred during calculation.")
        return None


num = float(input("Enter a number: "))

result = calculate(num)
print("Result:", result)
```
