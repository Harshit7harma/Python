```python
#Problem 1: Create a DataFrame to display the given result
import pandas as pd
data = {
    'Student': ['David', 'Samuel', 'Terry', 'Evan'],
    'Age': [27, 24, 22, 32],
    'Country': ['UK', 'Canada', 'China', 'USA'],
    'Course': ['Python', 'Data Structures',
               'Machine Learning', 'Web Development'],
    'Marks': [85, 72, 89, 76]
}
df = pd.DataFrame(data)
print(df)
```

```python
#Problem 2: Retrieve the Marks column and assign it to variable b
b = df['Marks']
print(b)
```

```python
#Problem 3: Retrieve the Country and Course columns and assign them to variable c
c = df[['Country', 'Course']]
print(c)
```

```python
#Problem 4: Data Loading & Exploration
import pandas as pd
data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan',
             'Neha', 'Arjun', 'Simran', 'Rohit', 'Anjali'],
    'Age': [25, 32, 40, 28, 35, 45, 22, 31, 38, 29],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],
    'Purchase_Amount': [2000, 3500, 4000, 2500, 5000,
                        4500, 1500, 3000, 5500, 2800]
}
df = pd.DataFrame(data)
# Display first 5 rows
print(df.head())
# Display summary statistics
print(df.describe())
# Check missing values
print(df.isnull().sum())
# Handle missing values
df = df.ffill()
print(df)
```

```python
#Problem 5: Data Filtering and Selection
#A: Retrieve all customers whose Age is greater than 30
print(df[df['Age'] > 30])
```

```python
#Problem 5: Data Filtering and Selection
#B Find the total Purchase Amount for male and female customers separately
print(df.groupby('Gender')['Purchase_Amount'].sum())
```

```python
#Problem 5: Data Filtering and Selection
#C Select only Customer_ID, Name, and Purchase_Amount
print(df[['Customer_ID', 'Name', 'Purchase_Amount']])
```

```python
#Problem 6: Data Transformation and Aggregation
#A Add a new column Discounted_Price with a 10% discount
df['Discounted_Price'] = df['Purchase_Amount'] * 0.90
print(df)
```

```python
#Problem 6: Data Transformation and Aggregation
#B Group by Gender and calculate average Purchase Amount
print(df.groupby('Gender')['Purchase_Amount'].mean())
```

```python
#Problem 6: Data Transformation and Aggregation
#C Sort the dataset by Purchase Amount in descending order
df = df.sort_values(
    by='Purchase_Amount',
    ascending=False
)
print(df)
```
